```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

num_records = 100  # adjust as needed

data = []

for _ in range(num_records):
    record = {
        # Timestamp (Real-time tracking)
        "timestamp": datetime.now() - timedelta(minutes=np.random.randint(0, 1440)),

        # Asset / Shipment Info
        "asset_id": f"TRUCK{np.random.randint(100, 999)}",
        "shipment_id": f"SHP{np.random.randint(1000, 9999)}",

        # GPS Sensor
        "latitude": round(np.random.uniform(35.0, 36.0), 6),
        "longitude": round(np.random.uniform(139.0, 140.0), 6),

        # Shipment Status
        "shipment_status": np.random.choice([
            "In Transit", "Delivered", "Delayed", "Out for Delivery"
        ]),

        # Inventory & Operations
        "inventory_level": np.random.randint(10, 500),
        "waiting_time": np.random.randint(0, 120),

        # IoT Sensors
        "temperature": round(np.random.uniform(-5.0, 25.0), 1),
        "humidity": np.random.randint(30, 90),

        # Traffic Condition
        "traffic_status": np.random.choice([
            "Clear", "Moderate", "Heavy", "Detour"
        ]),

        # Demand & Utilization
        "demand_forecast": np.random.randint(50, 300),
        "asset_utilization": round(np.random.uniform(50, 100), 2),

        # Delay Reason
        "logistics_delay_reason": np.random.choice([
            "Weather", "Traffic", "Mechanical Failure", "None"
        ]),

        # Target Variable (ML)
        "logistics_delay": np.random.choice([0, 1]),

        # RFID (Identity verification)
        "rfid_tag": f"RFID{np.random.randint(100000, 999999)}",
        "rfid_verified": np.random.choice([True, False]),

        # Fraud Prevention
        "tamper_alert": np.random.choice(["Yes", "No"])
    }

    data.append(record)

# Convert to DataFrame
df = pd.DataFrame(data)

# Save dataset
df.to_csv("iot_data.csv", index=False)
df.to_json("iot_data.json", orient="records",  date_format="iso")

# Preview
print(df.head())
```

```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

num_records = 100

yamato_statuses = [
    "Departure Scan",
    "In Transit",
    "Arrival Scan",
    "Out for Delivery",
    "Delivered",
    "Hold at Yamato",
    "Under Investigation",
    "Returned",
    "Delay"
]

japan_post_statuses = [
    "Arrival",
    "Delivered to the delivery address",
    "Hand it over at the window",
    "Bring it back due to your absence",
    "Storage",
    "Returned to the sender"
]

locations = [
    "Tokyo Central Post Office",
    "Osaka Central Post Office",
    "Nagoya Central Post Office",
    "Sapporo Central Post Office",
    "Naha Central Post Office",
    "Yokohama Sales Office",
    "Fukuoka Distribution Center"
]

prefectures = [
    "Tokyo",
    "Osaka",
    "Aichi",
    "Hokkaido",
    "Okinawa",
    "Kanagawa",
    "Fukuoka"
]

delay_reasons = [
    "None",
    "Weather",
    "Traffic",
    "Mechanical Failure",
    "Address Unknown",
    "Recipient Not Available"
]

data = []

for _ in range(num_records):
    carrier = np.random.choice(["Yamato Transport", "Japan Post"])

    if carrier == "Yamato Transport":
        latest_status = np.random.choice(yamato_statuses)
        tracking_number = str(np.random.randint(100000000000, 999999999999))
    else:
        latest_status = np.random.choice(japan_post_statuses)
        tracking_number = str(np.random.randint(10000000000, 999999999999))

    is_delay = 1 if latest_status in [
        "Delay",
        "Under Investigation",
        "Returned",
        "Bring it back due to your absence",
        "Storage"
    ] else 0

    record = {
        # Basic tracking data
        "timestamp": datetime.now() - timedelta(minutes=np.random.randint(0, 1440)),
        "carrier": carrier,
        "tracking_number": tracking_number,
        "package_id": f"PKG{np.random.randint(1000, 9999)}",

        # Shipment movement
        "origin": np.random.choice(["Tokyo", "Osaka", "Nagoya", "Fukuoka"]),
        "current_location": np.random.choice(locations),
        "delivery_location": np.random.choice(["Tokyo", "Osaka", "Kyoto", "Yokohama", "Sapporo"]),
        "prefecture": np.random.choice(prefectures),

        # GPS tracking
        "latitude": round(np.random.uniform(35.0, 36.0), 6),
        "longitude": round(np.random.uniform(139.0, 140.0), 6),

        # Tracking status
        "latest_status": latest_status,
        "logistics_delay_reason": np.random.choice(delay_reasons) if is_delay == 1 else "None",
        "logistics_delay": is_delay,

        # Delivery details
        "order_date": datetime.now() - timedelta(days=np.random.randint(1, 7)),
        "expected_delivery_date": datetime.now() + timedelta(days=np.random.randint(1, 5)),
        "waiting_time_minutes": np.random.randint(0, 180),

        # IoT sensor data
        "perishable": np.random.choice(["Yes", "No"]),
        "temperature": round(np.random.uniform(-5.0, 25.0), 1),
        "humidity": np.random.randint(30, 90),

        # RFID and fraud prevention
        "rfid_tag": f"RFID{np.random.randint(100000, 999999)}",
        "rfid_verified": np.random.choice([True, False]),
        "tamper_alert": np.random.choice(["Yes", "No"]),

        # Smart logistics data
        "traffic_status": np.random.choice(["Clear", "Moderate", "Heavy", "Detour"]),
        "inventory_level": np.random.randint(10, 500),
        "asset_utilization": round(np.random.uniform(50, 100), 2)
    }

    data.append(record)

df = pd.DataFrame(data)

df.to_csv("smart_logistic_tracker_japan.csv", index=False)

df.to_json(
    "smart_logistic_tracker_japan.json",
    orient="records",
    date_format="iso"
)

df.head()
```
